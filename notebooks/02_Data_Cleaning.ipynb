# Notebook 02 – Data Cleaning

## Objective

Clean the raw HHS dataset before Exploratory Data Analysis (EDA) and forecasting.

Tasks

- Remove completely empty rows
- Remove duplicate rows
- Convert data types
- Save cleaned dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/HHS_Unaccompanied_Alien_Children_Program.csv")

df.head()
```

```python
df = df.dropna(how="all")
```

```python
df.shape
```

## Step 2 - Verify Cleaning

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

## Step 3 - Convert Date Column

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.dtypes
```

## Step 4 - Convert Children in HHS Care to Numeric

```python
df["Children in HHS Care"] = (
    df["Children in HHS Care"]
    .str.replace(",", "", regex=False)
    .astype(int)
)
```

```python
df.dtypes
```

```python
df.head()
```

### Business Conclusion

The "Children in HHS Care" column was originally stored as text because the values contained thousands separators (commas). These commas were removed and the column was converted into a numeric data type so that it can be used for statistical analysis, visualization, and machine learning forecasting models.

## Step 5 - Save Cleaned Dataset

```python
df.to_csv(
    "../data/processed/HHS_UAC_Cleaned.csv",
    index=False
)
```
